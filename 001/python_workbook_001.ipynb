## 使い方

1. 上部の「ドライブにコピー」をクリックして、自分の複製を作ります
2. 書籍の問題を読み、各問題の解答セルにコードを書きます
3. 解答セルを実行してから、そのすぐ下の検証セルを実行します
4. `OK` と表示されれば正解です

検証セルの中身は、今の段階では分からなくて構いません。
書き換えずにそのまま実行してください。

解答を書き直したときは、必ず解答セルを実行し直してから検証セルを実行しましょう。
