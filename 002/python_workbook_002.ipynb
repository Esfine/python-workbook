## 使い方

1. 上部の「ドライブにコピー」をクリックして、自分の複製を作ります
2. 書籍の問題を読み、各問題の解答セルにコードを書きます
3. 解答セルを実行してから、そのすぐ下の検証セルを実行します
4. `OK` と表示されれば正解です

検証セルの中身は、今の段階では分からなくて構いません。
書き換えずにそのまま実行してください。

解答を書き直したときは、必ず解答セルを実行し直してから検証セルを実行しましょう。

## 第 8 章：総合問題

### 問題 8-1（応用）

```python
# ここに解答を書きます
```

```python
expected_pay = {10: 12600, 9: 11100, 8: 9600, 6: 7200}
expected_overtime = {10: 2, 9: 1, 8: 0, 6: 0}
assert hours in expected_pay, f"hours を {list(expected_pay)} のいずれかにしてください"
assert overtime == expected_overtime[hours], f"hours={hours} のとき overtime が {overtime} です（期待値: {expected_overtime[hours]}）"
assert pay == expected_pay[hours], f"hours={hours} のとき pay が {pay} です（期待値: {expected_pay[hours]}）"

expected = f"日給は {expected_pay[hours]:,} 円です（残業 {expected_overtime[hours]} 時間）"
assert message == expected, f"hours={hours} のとき message が「{message}」です（期待値:「{expected}」）"
print(f"hours={hours}: OK")
```
